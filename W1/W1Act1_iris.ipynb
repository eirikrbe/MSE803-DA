```python
from ucimlrepo import fetch_ucirepo 
import seaborn as sns
  
# fetch dataset 
iris = fetch_ucirepo(id=53) 
  
# data (as pandas dataframes) 
X = iris.data.features 
y = iris.data.targets 
  
# metadata 
print(iris.metadata) 
  
# variable information 
print(iris.variables) 
```

```python
y['class'].value_counts()
```

```python
# Inspect feature columns and summary statistics for X
X.info()
X.describe()
```

```python
# Combine predictors and target into a single DataFrame for grouped analysis
iris_df = X.copy()
iris_df['class'] = y['class']
iris_df.groupby('class').describe()
```

```python
# Count features, classes, and duplicate full records in the combined iris_df
n_features = X.shape[1]
n_classes = y['class'].nunique()

n_duplicates = iris_df.duplicated().sum()

print('Features:', n_features)
print('Classes:', n_classes)
print('Duplicate rows in iris_df:', n_duplicates)

if n_duplicates:
    display(iris_df[iris_df.duplicated(keep=False)])
```

```python
import matplotlib.pyplot as plt

# Plot pairwise relationships across all features and color by class.
g = sns.pairplot(iris_df, hue='class', corner=True)
g.savefig('iris_pairplot.png', dpi=200)
plt.show()
```
